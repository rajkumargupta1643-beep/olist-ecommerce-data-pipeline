```python
import requests, json

API_BASE_URL = "http://mock-api:8000"

# 1. Login karke token lo
resp = requests.post(f"{API_BASE_URL}/api/v1/auth/token",
                     json={"username": "candidate", "password": "blue-owls-2026"})
print("Login status:", resp.status_code)
print("Login response keys:", list(resp.json().keys()))
token = resp.json()["access_token"]
headers = {"Authorization": f"Bearer {token}"}

# 2. Har endpoint ka schema dekho (login ki zaroorat nahi)
for ep in ["orders", "order_items", "customers", "products", "sellers", "payments"]:
    r = requests.get(f"{API_BASE_URL}/api/v1/data/schema/{ep}")
    print("\n=====", ep, "=====")
    print(json.dumps(r.json(), indent=2)[:1500])

# 3. Orders ka ek chhota page lo
r = requests.get(f"{API_BASE_URL}/api/v1/data/orders",
                 headers=headers, params={"page": 1, "page_size": 3})
print("\n===== ORDERS PAGE 1 =====")
print("Status:", r.status_code)
if r.status_code == 200:
    body = r.json()
    print("Response type:", type(body).__name__)
    print(json.dumps(body, indent=2)[:2500])
else:
    print("Error body:", r.text)
```
